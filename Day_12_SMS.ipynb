# Real-Time Example: Student Management System
Scenario

A college wants to store student information.

We will use:

List → Subjects selected by the student
Tuple → Date of Birth (fixed, cannot change)
Set → Skills (duplicates not allowed)
Dictionary → Student Details

```python
# students = []

print("=" *60)
print("         Student Management System")
print("=" *60)

while True:
    student = {}
    print("\n Enter Student Details")
    student["ID"] = int(input("Student ID : "))
    student["Name"] = input("Student Name :")
    student["Age"] = input("Student Age :")
    student["city"] = input("Student city :")

    subjects = []

    print("\n Enter the Subjects")
    while True:
        subject = input("Subject : ")
        subjects.append(subject)
        choice = input("Would you like to add another subject (Y/N):")
        if choice.upper() == "N":
            break
    student["Subjects"] = subjects

    print("\n Enter the data of Birth")
    day = int(input("Day : "))
    month = input("Month :")
    year = int(input("Year :"))
    #dob = ()
    dob = (day,month,year)

    student["DOB"] = dob

    print("\n Enter the Skills")

    Skills = set()

    while True:
        skill = input("Enter the skill")
        Skills.add(skill)
        choice = input("Would you like to add another Skill (Y/N):")
        if choice.upper() == "N":
            break
    student["Skills"] = Skills

    students.append(student)
    choice = input("Would you like to add another Student (Y/N):")
    if choice.upper() == "N":
        break

####################################################
# Display Studnet informaiton
########################################

print("\n")
print("=" *70)
print("All Studnet informaiton")
print("=" *70)

for student in students:
    print("\n ________________________________________________________")
    print(f"ID is :{student['ID']}")
    print(f"Name is :{student['Name']}")
    print(f"Age is :{student['Age']}")
    print(f"City is :{student['city']}")
    print("\n DOB")

    dob = student['DOB']

    print(f"{dob[0]}-{dob[1]}-{dob[2]}")

    print("\n subjects enrolled")

    for subject in student["Subjects"]:
        print(subject)

    print("\n Skills are:")

    for skill in student["Skills"]:
        print(skill)

    print("____________________________________________________________")
    



    

    


    
    
        
        

    

    
              

    
            
        
        
        
    
    

```
