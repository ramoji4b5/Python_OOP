```python
class InvalidAgeError(Exception):
    print("Student should be 5 year old")



    
```

```python
try:
    s_age = int(input("enter Student age  :"))
    if s_age <=5:
        raise InvalidAgeError       
except InvalidAgeError:
    print("Student must the 5 years old")
```

```python
class student:
    def __init__(self,name,school,s_id,s_age):
        self.name = name
        self.school = school
        self.s_id= s_id
        self.s_age = s_age
        
    def display(self):
        print(f"student nam is {self.name}")
        print(f"student age is {self.s_age}")
        print(f"student id is {self.s_id}")
        print(f"school name is {self.school}")
        
```

```python
class Student(BaseModel):
    sid:int
    sname:str
    age:int
    school_name:str

    def display_one(self):
        print(f"student nam is {self.sname}")
        print(f"student age is {self.age}")
        print(f"student id is {self.sid}")
        print(f"school name is {self.school_name}")
```

```python
from pydantic import ValidationError
try:
    std2 = Student(sid=101,sname="ramoji",age="Twenty",school_name="VPS")
except ValidationError as exc:
    print(repr(exc.errors()))
    print("there is data validation error")
```

```python
try:
    std2 = Student(sid=101,sname="ramoji",age=20,school_name="VPS")
except ValidationError as exc:
    print(repr(exc.errors()))
    print("there is data validation error")
```

```python
std2.display_one()
```

```python
from pydantic import BaseModel


class Student(BaseModel):
    sid: int
    sname: str
    age: int
    school_name: str

    def display_two(self):
        print(f"Student Name  : {self.sname}")
        print(f"Student Age   : {self.age}")
        print(f"Student ID    : {self.sid}")
        print(f"School Name   : {self.school_name}")


# Create Object
std2 = Student(
    sid=101,
    sname="John",
    age=20,
    school_name="ABC Public School"
)

# Call Method
std2.display_two()
```
