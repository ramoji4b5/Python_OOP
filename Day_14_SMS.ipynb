```python
def register_student():
    print("\n--- Register Student ---")

    sid = int(input("ID: "))
    if find_student(sid):
        print("Student ID already exists.")
        return

    student = {}
    student["ID"] = sid
    student["Name"] = input("Name: ")
    student["Age"] = int(input("Age: "))
    student["City"] = input("City: ")

    subjects = []
    print("\nEnter Subjects")
    while True:
        subjects.append(input("Subject: "))
        if input("Add another subject (Y/N): ").upper() == "N":
            break
    student["Subjects"] = subjects

    print("\nEnter DOB")
    day = int(input("Day: "))
    month = input("Month: ")
    year = int(input("Year: "))
    student["DOB"] = (day, month, year)

    skills = set()
    print("\nEnter Skills")
    while True:
        skills.add(input("Skill: "))
        if input("Add another skill (Y/N): ").upper() == "N":
            break
    student["Skills"] = skills

    students.append(student)
    print("Student registered successfully.")
```

```python
"""
student_management.py
A beginner-friendly Student Management System demonstrating:
- List, Tuple, Set, Dictionary
- Functions
- Loops
- Menu-driven programming
"""

#students = []


def find_student(student_id):
    for student in students:
        if student["ID"] == student_id:
            return student
    return None
```

```python
def display_students():
    print("\n========== STUDENTS ==========")
    if not students:
        print("No students found.")
        return

    for s in students:
        print("-" * 40)
        print(f"ID      : {s['ID']}")
        print(f"Name    : {s['Name']}")
        print(f"Age     : {s['Age']}")
        print(f"City    : {s['City']}")
        d = s["DOB"]
        print(f"DOB     : {d[0]}-{d[1]}-{d[2]}")
        print("Subjects:", ", ".join(s["Subjects"]))
        print("Skills  :", ", ".join(sorted(s["Skills"])))
```

```python
def search_student():
    sid = int(input("Enter Student ID: "))
    s = find_student(sid)
    if s:
        print("\nStudent Found")
        print(s)
    else:
        print("Student not found.")
```

```python
def update_student():
    sid = int(input("Enter Student ID: "))
    s = find_student(sid)

    if not s:
        print("Student not found.")
        return

    while True:
        print("""
            1.Name
            2.Age
            3.City
            4.Subjects
            5.Skills
            6.Back
            """)
        ch = input("Choice: ")

        if ch == "1":
            s["Name"] = input("New Name: ")

        elif ch == "2":
            s["Age"] = int(input("New Age: "))

        elif ch == "3":
            s["City"] = input("New City: ")

        elif ch == "4":
            while True:
                print("""
                        Subjects Menu
                        1.Add
                        2.Remove
                        3.Display
                        4.Back
                        """)
                op = input("Choice: ")
                if op == "1":
                    s["Subjects"].append(input("Subject: "))
                elif op == "2":
                    sub = input("Subject to remove: ")
                    if sub in s["Subjects"]:
                        s["Subjects"].remove(sub)
                elif op == "3":
                    for sub in s["Subjects"]:
                        print(sub)
                else:
                    break

        elif ch == "5":
            while True:
                print("""
                        Skills Menu
                        1.Add
                        2.Remove
                        3.Display
                        4.Back
                        """)
                op = input("Choice: ")
                if op == "1":
                    s["Skills"].add(input("Skill: "))
                elif op == "2":
                    s["Skills"].discard(input("Skill to remove: "))
                elif op == "3":
                    for skill in sorted(s["Skills"]):
                        print(skill)
                else:
                    break
        else:
            break
```

```python

def delete_student():
    sid = int(input("Enter Student ID: "))
    s = find_student(sid)
    if s:
        students.remove(s)
        print("Student deleted.")
    else:
        print("Student not found.")
```

```python
def subject_report():
    print("\n===== SUBJECT REPORT =====")
    subject_map = {}

    for s in students:
        for sub in s["Subjects"]:
            subject_map.setdefault(sub.lower(), []).append(s["Name"])

    if not subject_map:
        print("No subjects available.")
        return

    for sub, names in sorted(subject_map.items()):
        print(f"\n{sub}")
        for n in names:
            print(" -", n)
```

```python
subject_report()
```

```python
def unique_skills():
    print("\n===== UNIQUE SKILLS =====")
    all_skills = set()

    for s in students:
        all_skills.update(s["Skills"])

    for skill in sorted(all_skills):
        print(skill)
```

```python
def statistics():
    print("\n===== STATISTICS =====")

    print("Total Students :", len(students))

    if students:
        avg = sum(s["Age"] for s in students) / len(students)
        print("Average Age    :", round(avg, 2))

    cities = {s["City"] for s in students}
    print("Cities         :", ", ".join(sorted(cities)))

    total_subjects = sum(len(s["Subjects"]) for s in students)
    print("Total Subjects :", total_subjects)

    all_skills = set()
    for s in students:
        all_skills.update(s["Skills"])
    print("Unique Skills  :", len(all_skills))
```

```python
def main():
    while True:
        print("""
                ===============================
                 STUDENT MANAGEMENT SYSTEM
                ===============================
                1.Register Student
                2.Display Students
                3.Search Student
                4.Update Student
                5.Delete Student
                6.Subject Report
                7.Unique Skills
                8.Statistics
                9.Exit
                """)
        choice = input("Enter Choice: ")

        if choice == "1":
            register_student()
        elif choice == "2":
            display_students()
        elif choice == "3":
            search_student()
        elif choice == "4":
            update_student()
        elif choice == "5":
            delete_student()
        elif choice == "6":
            subject_report()
        elif choice == "7":
            unique_skills()
        elif choice == "8":
            statistics()
        elif choice == "9":
            print("Thank you!")
            break
        else:
            print("Invalid choice.")


if __name__ == "__main__":
    main()
```
