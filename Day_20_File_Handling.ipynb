```python
from pydantic import BaseModel

class Student(BaseModel):
    sid: int
    sname: str
    age: int
    school_name: str

    def display_two(self):
        print(f"Student Name  : {self.sname}")
        print(f"Student Age   : {self.age}")
        print(f"Student ID    : {self.sid}")
        print(f"School Name   : {self.school_name}")


# Create Object
std2 = Student(
    sid=101,
    sname="John",
    age=20,
    school_name="ABC Public School"
)

# Call Method
```

```python
student.append(std2)
```

# with to create the file

```python
with open("students.txt","r") as file3:
    data = file3.read()
    print(data)
```

```python
with open("students.txt","r") as file3:
    data = file3.readline()
    print(data)
```

```python
with open("students.txt","r") as file3:
    data = file3.readlines()
    print(data)
```

```python
def save_student(std2):
    with open("students.txt","a") as file:
        file.write(f"{std2.sid},"
                   f"{std2.age},"
                   f"{std2.sname},"
                   f"{std2.school_name}\n"
        )
        
```

```python
save_student(std2)
```

```python
std3 = Student(
    sid=103,
    sname="Kumar",
    age=25,
    school_name="VPS Public School"
)
```

```python
save_student(std3)
```
