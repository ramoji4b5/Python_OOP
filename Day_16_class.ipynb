# One School --> Many Students

```python
class student:
    def __init__(self,sid,sname):
        self.sid = sid
        self.sname = sname

    def display(self):
        print(f"ID is :{self.sid}")
        print(f"Name is :{self.sname}")
        print(f"_" *40)

class school:
    def __init__(self,school_name):
        self.school_name = school_name
        self.students = []   # Student List  #SCHOOL HAS MANY Student

    def add_student(self,student):
        self.students.append(student)

    def display_students(self):
        print(f" School name is {self.school_name}")
        print(f"=" * 40)
        for student in self.students:
            student.display()
        
        
    
```

# Using both IS_A & HAS_A

```python
# Parent
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
    def display(self):
        print(f"Name is :{self.name}")
        print(f"age is :{self.age}")
        
# Child Claas( IS-A)
class Student(Person):
    def __init__(self,std_id,name,age):
        super().__init__(name,age)
        self.std_id=std_id

    def display_std_info(self):
        print(f"Student ID Is :{self.std_id}")

# School (HAS-A)
class School:
    def __init__(self,school_name):
        self.school_name = school_name
        self.students = [] # HAS A (MANY STUDENTS)

    def add_student(self,student):
        self.students.append(student)

    def display_students(self):
        print(f" School name is {self.school_name}")
        print(f"=" * 40)
        for student in self.students:
            student.display()
            student.display_std_info()
            

    
    


    
        
    
```

```python
school = School("Varma Public School")
```

```python
student1 = Student(101,"Ramoji",36)
student2 = Student(102,"Lakshmoji",34)
student3 = Student(103,"Ganeshji",33)
```

```python
school.add_student(student1)
school.add_student(student2)
school.add_student(student3)
```
