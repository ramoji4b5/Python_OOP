```python
class BankAccountInfo:
    def __init__(self,account_no,name,age,address,phone,balance):
        self.account_no = account_no
        self.name = name
        self.age = age
        self.address = address
        self.phone = phone
        self.balance = balance

    def display(self):
        print("\n Customer Information")
        print("=" * 40)

        details = {
            "Account Number" : self.account_no,
            "Name" : self.name,
            "Age" : self.age,
            "Address" : self.address,
            "Phone no" : self.phone,
            "Balance" : f"${self.balance:.2f}"
            }
            
        for key,value in details.items():
            print(f"{key:<18}:{value}")

    def deposit(self,amount):
        self.balance += amount

        print("\n Deposit Successful")
        print(f"Desposited Amount is : ${amount:.2f}")
        print(f" Total Balance is : ${self.balance:.2f}")

    def withdraw(self,amount):
        
        if amount <= self.balance:

            self.balance -= amount
            print("\n Withdraw is Successful")
            print(f"Withdraw Amount is : ${amount:.2f}")
            print(f" Total Balance is : ${self.balance:.2f}")
        else:
            print("\n Insufficient balance")

    def show_balance(self):
        print(f"\n Avilable balance is : ${self.balance:.2f}")


customers = {
    101 : BankAccountInfo(101,"Ramoji",35,"Hyderabad","9494021190",25000),
    102 : BankAccountInfo(102,"Kasi",32,"vijayawada","9497612345",5000),
    103 : BankAccountInfo(103,"Rani",26,"guntur","93423452345",35000),
    104 : BankAccountInfo(104,"Sita",27,"kadapa","92345234523",1000),
}


while True:

    print("\n")
    print("$" * 50)
    print("            ABC Bank Management System")
    print("#" * 45)
    print(" 1. Display customer Details")
    print(" 2. checking Balance")
    print(" 3. Desposit Money")
    print(" 4. Withdraw Money")
    print(" 5. Display all customer")
    print(" 6. Exit")

    choice = input("\n Enter Your choice :")

    if choice not in("1","2","3","4","5","6"):
        print("\n Invalid Menu choice")
        continue

    if choice == "5":
        print("\n customer List")
        print("-" * 40)
        for account_no,customer in customers.items():
            print(
                f"{account_no: <10}"
                f"{customer.name:<15}"
                f"{customer.balance:.2f}"
            )
        continue

    if choice == "6":
        print("\n Thank you for banking with ABC bank")
        print("VISIT AGAIN!")
        break

    account = int(input("\n Enter the Account Number:"))
    
    if account not in customers:
        print("\n Customer not found")
        continue
        
    customer = customers[account]
    if choice == "1":
        customer.display()
    elif choice == "2":
        customer.show_balance()
    elif choice == "3":
        amount = float(input("Enter the Deposit amount :"))
        if amount >= 0:
            customer.deposit(amount)
        else:
            print("Desposit should be grater than zero")
        continue
    elif choice == "4":
        amount = float(input("Enter the Withdraw amount :"))
        if amount <=0:
            print("Invalid wihtdraw amount")
            continue
        customer.withdraw(amount)
        
        
    
        
        
        
    
        
    
    
    
    

    
    

        
        
        
            
              
            
        
        
        
        
```
